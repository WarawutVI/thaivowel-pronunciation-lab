```python
import pandas as pd

# ระบุพาธไปยังไฟล์ CSV ใน Google Drive ของคุณ
file_path = '/content/drive/MyDrive/Colab Notebooks/เเบบประเมินความพึ่งพอใจของผู้ใช้งาน .csv'

# โหลดข้อมูลเข้าสู่ DataFrame
df = pd.read_csv(file_path)

# แสดงข้อมูล 5 แถวแรก
display(df.head())
```

```python
# รายชื่อคอลัมน์ประเมินผลทั้งหมด (คอลัมน์ที่ 7 ถึง 19)
evaluation_cols = [
    '2.1  ด้านการใช้งานและการออกแบบ (Usability & UI Design) [ใช้งานง่าย ไม่ซับซ้อน]',
    '2.1  ด้านการใช้งานและการออกแบบ (Usability & UI Design) [มีความสวยงามและเป็นระเบียบ]',
    '2.1  ด้านการใช้งานและการออกแบบ (Usability & UI Design) [การแสดงผลสองภาษา (ไทย/อังกฤษ) ชัดเจน]',
    '2.1  ด้านการใช้งานและการออกแบบ (Usability & UI Design) [รูปเเบบตัวอักษร อ่านง่ายเเละสวยงาม]',
    '2.2 ด้านความถูกต้องและประสิทธิภาพของระบบ (System Accuracy & Performance) [ความถูกต้องครบถ้วนของข้อมูล]',
    '2.2 ด้านความถูกต้องและประสิทธิภาพของระบบ (System Accuracy & Performance) [จำแนกเสียงสระได้ถูกต้องแม่นยำ]',
    '2.2 ด้านความถูกต้องและประสิทธิภาพของระบบ (System Accuracy & Performance) [ผลป้อนกลับ (Feedback) ชัดเจน เข้าใจง่าย]',
    '2.2 ด้านความถูกต้องและประสิทธิภาพของระบบ (System Accuracy & Performance) [ประมวลผลและแสดงผลได้รวดเร็ว]',
    '2.3 ด้านประโยชน์ต่อการเรียนรู้ (Learning Effectiveness) [ช่วยให้ออกเสียงสระภาษาไทยได้ถูกต้องขึ้น]',
    '2.3 ด้านประโยชน์ต่อการเรียนรู้ (Learning Effectiveness) [ช่วยเพิ่มความมั่นใจในการออกเสียง]',
    '2.3 ด้านประโยชน์ต่อการเรียนรู้ (Learning Effectiveness) [สร้างแรงจูงใจให้ฝึกฝนอย่างต่อเนื่อง]',
    '2.3 ด้านประโยชน์ต่อการเรียนรู้ (Learning Effectiveness) [มีประโยชน์ต่อการเรียนรู้การออกเสียงภาษาไทย]',
    '2.3 ด้านประโยชน์ต่อการเรียนรู้ (Learning Effectiveness) [นำไปปรับใช้ในการออกเสียงในชีวิตประจำวัน]'
]

# วนลูปเพื่อหาความถี่ของแต่ละคำตอบแยกตามกลุ่มอายุในทุกคอลัมน์
for i, col in enumerate(evaluation_cols, 7):
    if col in df.columns:
        print(f"\n--- ตารางที่ {i}: {col} ---")
        ct = pd.crosstab(df['กลุ่มอายุ_ใหม่'], df[col])
        display(ct)
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

results = []

for col in evaluation_cols:
    if col in df.columns:
        # สร้าง Contingency Table ระหว่างกลุ่มอายุกับข้อคำถามนั้นๆ
        contingency_table = pd.crosstab(df['กลุ่มอายุ_ใหม่'], df[col])

        try:
            chi2, p_val, dof, expected = chi2_contingency(contingency_table)
            results.append({
                'ข้อประเมิน': col,
                'Chi-Square': round(chi2, 4),
                'p-value': round(p_val, 4),
                'd.f.': dof,
                'มีนัยสำคัญ (p < 0.05)': 'ใช่' if p_val < 0.05 else 'ไม่ใช่'
            })
        except Exception as e:
            results.append({
                'ข้อประเมิน': col,
                'Chi-Square': None,
                'p-value': None,
                'd.f.': None,
                'มีนัยสำคัญ (p < 0.05)': f'ข้อผิดพลาด: {str(e)}'
            })

# แปลงเป็น DataFrame เพื่อการแสดงผลที่สวยงาม
chi_square_summary = pd.DataFrame(results)
display(chi_square_summary)
```
